```python
plt.plot(vf['Year'],vf['Kohli_Runs'],label="virat kholi runs",color="blue",linestyle="-")
plt.plot(vf['Year'],vf['Rohit_Runs'],label="rohit ke runs",color="red",linestyle="-")

plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Kohli vs Rohit Runs Over Years")
plt.legend()
plt.show()
```
